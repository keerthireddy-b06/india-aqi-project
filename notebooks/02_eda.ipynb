```python
#Load the processed file step
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
hyd = pd.read_csv( 
'../data/processed/hyderabad_aqi.csv', 
parse_dates=['Date'], 
index_col='Date' 
) 
```

```python
#  AQI over time
hyd['AQI'].plot(figsize=(14, 4), title='Hyderabad AQI over time') 
plt.ylabel('AQI') 
plt.show()
```

```python
#  Average AQI by month. 
monthly = hyd.assign(month=hyd.index.month).groupby('month')['AQI'].mean() 
monthly.plot(kind='bar', title='Average AQI by month') 
plt.show() 
```

```python
#   Pollutant correlations
plt.figure(figsize=(8, 6)) 
sns.heatmap(hyd.corr(numeric_only=True), annot=True, cmap='RdYlGn_r') 
plt.title('Correlation between AQI and pollutants') 
plt.show()
```
